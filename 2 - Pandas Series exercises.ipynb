## Series indexation

### Given the X pandas Series, show its first element


```python
# your code goes here
```

```python
# Create a Series with custom index names
X = pd.Series(['A','B','C'], index=['first', 'second', 'third'])

# Three ways to access the first element:
#X[0]           # by position (doesn't work with custom indices)
#X.iloc[0]      # by position (iloc = integer location)
X['first']      # by index label (preferred when labels exist)
```

### Given the X pandas Series, show its last element


```python
# your code goes here
```

```python
# Create a Series with custom index names
X = pd.Series(['A','B','C'], index=['first', 'second', 'third'])

# Three ways to access the last element:
#X[-1]          # by position (doesn't work with custom indices)
#X.iloc[-1]     # by position (iloc = integer location)
X['third']      # by index label (the preferred way)
```

![green-divider](https://user-images.githubusercontent.com/7065401/52071924-c003ad80-2562-11e9-8297-1c6595f8a7ff.png)

### Given the X pandas Series, show all middle elements


```python
# your code goes here
```

```python
# Create a Series with 5 elements and custom index names
X = pd.Series(['A','B','C','D','E'],
              index=['first','second','third','forth','fifth'])

# Three ways to get the middle elements (excluding first and last):
#X[['second', 'third', 'forth']]   # select by index labels
#X.iloc[1:-1]                      # by position using integer location
X[1:-1]                             # by position (simpler syntax)
```

![green-divider](https://user-images.githubusercontent.com/7065401/52071924-c003ad80-2562-11e9-8297-1c6595f8a7ff.png)

```python
# Create a Series with 5 elements and custom index names
X = pd.Series(['A','B','C','D','E'],
              index=['first','second','third','forth','fifth'])

# Two ways to reverse the Series order:
#X.iloc[::-1]    # by position using negative step
X[::-1]          # by label with negative step (simpler)
```

### Given the X pandas Series, show the first and last elements


```python
# your code goes here
```

```python
# Create a Series with 5 elements
X = pd.Series(['A','B','C','D','E'],
              index=['first','second','third','forth','fifth'])

# Three ways to get the first and last elements:
#X[['first', 'fifth']]        # select by index labels
#X.iloc[[0, -1]]              # by position using integer location
X[[0, -1]]                     # by position (can also use with labels)
```

## Series manipulation

### Convert the given integer pandas Series to float


```python
# your code goes here
```

```python
# Create a Series with integer values
X = pd.Series([1,2,3,4,5],
              index=['first','second','third','forth','fifth'])

# Convert integer Series to float Series
# The dtype parameter specifies the data type
pd.Series(X, dtype=np.float)
```

![green-divider](https://user-images.githubusercontent.com/7065401/52071924-c003ad80-2562-11e9-8297-1c6595f8a7ff.png)

### Reverse the given pandas Series (first element becomes last)

```python
# your code goes here
```

```python
# Create a Series with values in non-sequential order
X = pd.Series([1,2,3,4,5],
              index=['first','second','third','forth','fifth'])

# Reverse the Series using negative step ([::-1])
X[::-1]
```

![green-divider](https://user-images.githubusercontent.com/7065401/52071924-c003ad80-2562-11e9-8297-1c6595f8a7ff.png)

### Order (sort) the given pandas Series


```python
# your code goes here
```

```python
# Create a Series with unsorted values
X = pd.Series([4,2,5,1,3],
              index=['forth','second','fifth','first','third'])

# Sort the Series by values (not by index)
# sort_values() returns a new sorted Series
X = X.sort_values()
X
```
