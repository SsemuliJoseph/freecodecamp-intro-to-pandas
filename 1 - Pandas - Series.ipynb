## Hands on! 

```python
# Import the main pandas library - used for data manipulation and analysis
import pandas as pd

# Import numpy - provides mathematical functions and array operations
import numpy as np
```

## Pandas Series

We'll start analyzing "[The Group of Seven](https://en.wikipedia.org/wiki/Group_of_Seven)". Which is a political formed by Canada, France, Germany, Italy, Japan, the United Kingdom and the United States. We'll start by analyzing population, and for that, we'll use a `pandas.Series` object.

```python
# Create a pandas Series from a list of numbers
# A Series is like a 1D column of data, similar to a spreadsheet column
# These values represent the G7 countries' populations in millions
g7_pop = pd.Series([35.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523])
```

```python
g7_pop
```

Someone might not know we're representing population in millions of inhabitants. Series can have a `name`, to better document the purpose of the Series:

```python
# Give the Series a descriptive name
# This helps other people (and future you!) understand what the data represents
g7_pop.name = 'G7 Population in millions'
```

```python
# Display the Series to see what it looks like
# Notice: Each value has an automatic numeric index (0, 1, 2, etc.)
g7_pop
```

Series are pretty similar to numpy arrays:

```python
# Check the data type of values in the Series
# dtype tells us what type of data the Series contains (float64 = floating point numbers)
g7_pop.dtype
```

```python
# Access the underlying numpy array
# This shows the raw array without the index labels
g7_pop.values
```

They're actually backed by numpy arrays:

```python
# Check the type of the underlying values
# Series internally uses NumPy arrays for storage and computation
type(g7_pop.values)
```

And they _look_ like simple Python lists or Numpy Arrays. But they're actually more similar to Python `dict`s.

A Series has an `index`, that's similar to the automatic index assigned to Python's lists:

```python
# Display the Series to see the index
g7_pop
```

```python
# Access the first element using its position (0-based indexing)
# This is like accessing items in a list
g7_pop[0]
```

```python
# Access the second element using its position
# Notice: 1 is the position index (0 = first, 1 = second, etc.)
g7_pop[1]
```

```python
# Display the index (labels/row identifiers) of the Series
# This shows what labels are assigned to each value
g7_pop.index
```

```python
l = ['a', 'b', 'c']
```

But, in contrast to lists, we can explicitly define the index:

```python
# Assign meaningful labels (country names) to the index
# Now instead of accessing by position (0,1,2...), we can access by country name
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States',
]
```

```python
# Display the Series with the new country labels as index
# Now it's much clearer which population value belongs to which country
g7_pop
```

We can say that Series look like "ordered dictionaries". We can actually create Series out of dictionaries:

```python
# Alternative way: Create Series from a Python dictionary
# Dictionary keys automatically become the index labels
# This is more intuitive when you already have labeled data
pd.Series({
    'Canada': 35.467,
    'France': 63.951,
    'Germany': 80.94,
    'Italy': 60.665,
    'Japan': 127.061,
    'United Kingdom': 64.511,
    'United States': 318.523
}, name='G7 Population in millions')
```

```python
# Alternative way: Create Series by passing data and index separately
# This gives the same result as setting index after creation
pd.Series(
    [35.467, 63.951, 80.94, 60.665, 127.061, 64.511, 318.523],
    index=['Canada', 'France', 'Germany', 'Italy', 'Japan', 'United Kingdom',
       'United States'],
    name='G7 Population in millions')
```

You can also create Series out of other series, specifying indexes:

```python
# Create a new Series from an existing one, selecting only some countries
# Note: Spain is not in our data, so it will have NaN (missing value)
pd.Series(g7_pop, index=['France', 'Germany', 'Italy', 'Spain'])
```

## Indexing

Indexing works similarly to lists and dictionaries, you use the **index** of the element you're looking for:

```python
# Display the Series with country indices
g7_pop
```

```python
# Access a value using its label (country name) instead of position
# This is called label-based indexing and is more readable
g7_pop['Canada']
```

```python
# Access Japan's population using its label
# Much clearer than remembering that Japan is at position 4!
g7_pop['Japan']
```

Numeric positions can also be used, with the `iloc` attribute:

```python
# Access by position using .iloc[] (integer location)
# iloc[0] means the first row, regardless of its label
g7_pop.iloc[0]
```

```python
# Access the last element using -1 (negative indexing)
# -1 always refers to the last item, -2 to the second last, etc.
g7_pop.iloc[-1]
```

Selecting multiple elements at once:

```python
# Select multiple values at once using a list of labels
# Double square brackets: outer [] for the operation, inner [] for the list
g7_pop[['Italy', 'France']]
```

_(The result is another Series)_

```python
# Select multiple values using positions (the first two countries)
# This returns a new Series with those elements
g7_pop.iloc[[0, 1]]
```

Slicing also works, but **important**, in Pandas, the upper limit is also included:

```python
# Slice using labels - gets all countries from Canada to Italy (INCLUSIVE!)
# Important: Unlike Python lists, pandas includes the end label
g7_pop['Canada': 'Italy']
```

## Conditional selection (boolean arrays)

The same boolean array techniques we saw applied to numpy arrays can be used for Pandas `Series`:

```python
# Display the Series again
g7_pop
```

```python
# Create a boolean (True/False) mask for countries with population > 70 million
# This returns a Series of True/False values
g7_pop > 70
```

```python
# Use the boolean mask to filter: keep only countries with > 70 million people
# This is called boolean indexing
g7_pop[g7_pop > 70]
```

```python
# Calculate the average (mean) population of all G7 countries
g7_pop.mean()
```

```python
# Filter for countries with population greater than the average
# Combines comparison and boolean indexing
g7_pop[g7_pop > g7_pop.mean()]
```

```python
# Calculate standard deviation (measure of how spread out the data is)
g7_pop.std()
```

```python
# Boolean operators used in pandas:
# ~ means NOT (inverts True to False and vice versa)
# | means OR (at least one condition must be true)
# & means AND (all conditions must be true)
# Note: Use these instead of 'not', 'or', 'and' in pandas operations!
```

```python
# Complex filter: Find countries that are either:
# 1. Below (mean - std/2), OR
# 2. Above (mean + std/2)
# This selects extreme values (outliers)
g7_pop[(g7_pop > g7_pop.mean() - g7_pop.std() / 2) | (g7_pop > g7_pop.mean() + g7_pop.std() / 2)]
```

## Operations and methods
Series also support vectorized operations and aggregation functions as Numpy:

```python
# Display the Series
g7_pop
```

```python
# Vectorized operation: multiply all values by 1 million to convert to actual population numbers
# The operation is applied to every element without using a loop
g7_pop * 1_000_000
```

```python
# Calculate the mean (average) of all population values
g7_pop.mean()
```

```python
# Apply NumPy's natural logarithm function to each value in the Series
# This is another example of vectorized operations
np.log(g7_pop)
```

```python
# Slice and aggregate: calculate the mean of a subset of countries (France to Italy)
g7_pop['France': 'Italy'].mean()
```

## Boolean arrays
(Work in the same way as numpy)

```python
# Display the Series
g7_pop
```

```python
# Create a boolean mask: which countries have population > 80 million?
g7_pop > 80
```

```python
# Keep only countries with population > 80 million
g7_pop[g7_pop > 80]
```

```python
# Complex boolean filter: Find countries with population > 80 OR < 40 million
# This filters for extreme values at both ends
g7_pop[(g7_pop > 80) | (g7_pop < 40)]
```

```python
# Complex boolean filter: Find countries with population between 80 AND 200 million
g7_pop[(g7_pop > 80) & (g7_pop < 200)]
```

## Modifying series


```python
# Modify a single element by label: change Canada's population to 40.5
g7_pop['Canada'] = 40.5
```

```python
# Display the updated Series (notice Canada's new value)
g7_pop
```

```python
# Modify a single element by position: change the last country to 500
g7_pop.iloc[-1] = 500
```

```python
# Display the Series again (notice the last value changed)
g7_pop
```

```python
# Filter: Show only countries with population < 70 million
g7_pop[g7_pop < 70]
```

```python
# Bulk modification: Set all countries with < 70 million population to 99.99
# This is a conditional assignment - only values matching the condition are changed
g7_pop[g7_pop < 70] = 99.99
```

```python
# Display the final Series (all small countries are now set to 99.99)
g7_pop
```
